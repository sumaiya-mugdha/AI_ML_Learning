```python
import numpy as np

# 0-D Array (Scalar)
scalar_array = np.array(42)
print(f"0-D Array (Scalar):\n{scalar_array}")

# 1-D Array (Vector)
vector_array = np.array([1, 2, 3, 4, 5])
print(f"1-D Array (Vector):\n{vector_array}\n")

# 2-D Array (Matrix)
matrix_array = np.array([[10, 11, 12], [13, 14, 15]])
print(f"2-D Array (Matrix):\n{matrix_array}\n")
```

```python
# Creating random arrays
random_array_3d = np.random.rand(2, 2, 2)
print(f"3D Array with Random Numbers (2x2x2):\n{random_array_3d}")
```
