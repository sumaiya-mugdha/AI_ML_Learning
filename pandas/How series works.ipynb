```python
import pandas as pd
```

```python
s = pd.Series([10, -5, 0, 7, -2], index=["a","b","c","d","e"], name="Students")
s[s>0]
```

```python
#different series same size
s = pd.Series([10, -5, 0, 7, -2], index=["a","b","c","d","e"])
a = pd.Series([True, False, True, False, True], index=["a","b","c","d","e"])
s[a]
```

```python
s = pd.Series([10, -5, 0, 7, -2], index=["a", "b", "c", "d", "e"], name="Students")
a = pd.Series([-1, -2, 1, -9, -8], index=["a", "b", "c", "d", "e"])
#              f   f   T   f   f
s[a >= 0]
```

```python
s = pd.Series([10, -5, 0, 7, -2], index=["a", "b", "c", "d", "e"], name="Students")
a = pd.Series([-10, 75, 0, 7, -2], index=["tasik", "islam", "mug", "ron", "jhon"])
try:
    s[a]
except KeyError as error:
    print(error)
#GIVES ERROR AS INDEX IS DIFFERENT
#KeyError: "None of [Index([-10, 75, 0, 7, -2], dtype='int64')] are in the [index]"
```
