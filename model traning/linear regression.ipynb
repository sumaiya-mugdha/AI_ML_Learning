```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("D:/projects/AI_ML_LEARNING/placement (1).csv")
print(df.to_string())
```

```python
x= np.array(df["cgpa"])
y= np.array(df["iq"])
train_x= x[0: 90]
train_y= y[0: 90]
test_x= x[91:100]
test_y= y[91:100]
```

```python
# Plot the data points
plt.scatter(train_x, train_y, marker='x', c='r')
# Set the title
plt.title("CG VS IQ")
# Set the y-axis label
plt.ylabel('IQ')
plt.xlabel('CGPA')
# Set the x-axis label
plt.show()
```

```python
w = -25
b = 280
def compute_model_output(train_x, w, b):
    m = train_x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * train_x[i] + b

    return f_wb
tmp_f_wb = compute_model_output(train_x, w, b,)

# Plot our model prediction
plt.plot(train_x, tmp_f_wb, c='b',label='Our Prediction')

# Plot the data points
plt.scatter(train_x, train_y, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("CG VS IQ")
# Set the y-axis label
plt.ylabel('IQ')
# Set the x-axis label
plt.xlabel('CGPA')
plt.legend()
plt.show()
```
